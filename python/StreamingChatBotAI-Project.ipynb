```python
import sys
def get_yes_no(prompt):
    while True:
        answer = input(prompt).strip().lower()

        if answer in ("yes", "no"):
            return answer

        print("Please enter yes or no.")

# The Greeting & Main Menu
print ("Welcome to the Streaming Support Assistant")
print ("Lets get your device up and running")

# Asking the user to choose their main issue category
main_menu= input("What can I help you with today? 1. Power 2. Connection 3. Audio/Video 4. Remote (enter number 1-4)").strip()
# This runs if they choose "1" (Power Issue) from the main menu
if main_menu=="1":
  print ("Got it. Let's check the power")
  power_check= input ("Is the issue 1. Device won't turn on at all, or 2. Device keeps restarting?")
  if power_check=="1":
    print ("Let's figure out why it won't turn on.")
    light_check= get_yes_no("Is there any status light visible on the front of the device? (yes/no): ")
    if light_check=="yes":
      print ("Step: The device has power. Try switching your TV input source or checking your HDMI connection.")
      resolved= get_yes_no("Did switching the input fix the issue? (yes/no): ")
      if resolved=="no":
        print ("Escalation: Creating a ticket and connecting you to a Hardware Specialist for further diagnostics... ")
      else:
        print("Awesome! Glad it's working.")
    else:
      print ("Step: Ensure the power cable is securely plugged into both the wall outlet and the device.")
      resolved= get_yes_no("Did checking the cable plugs fix the issue? (yes/no): ")
      if resolved=="no":
        print ("Escalation: Creating a hardware replacement ticket for your power adapter...")
      else:
        print ("Great! You are good to go.")
  elif power_check=="2":
    print ("Let's tackle the constant restarting.")
    power_source= get_yes_no("Is the device plugged directly into a wall outlet? (yes/no): ")
    if power_source=="no":
      print ("Step: Move the power plug from the TV's USB port directly into a wall outlet using the power adapter.")
      resolved= get_yes_no("Did moving to a wall outlet stop the restarting? (yes/no): ")
      if resolved=="no":
        print ("Escalation: Creating a ticket and transferring to Tier 2 Support to check for a system boot loop...")
      else:
        print ("Perfect! Direct wall power solves a lot of issues.")
    else:
      print ("Step: Unplug the device, wait 60 seconds, and plug it back in to clear the system cache.")
      resolved= get_yes_no("Did the power cycle fix the issue? (yes/no): ")
      if resolved=="no":
        print ("Escalation: Creating ticket and connecting to Tier 2 Support to perform a factory reset...")
      else:
        print ("Excellent! Enjoy your streaming.")
  else:
    print ("Invalid choice for power troubleshooting. Please restart.")
# This runs if they choose "2" (Connection Issue) from the main menu
elif main_menu=="2":
  print ("Routing to Network Support")
  print ("Let's look at your connection...")
  conn_check= input("Are you seeing a 1. 'No Internet' error message, or 2. Is the video just buffering (spinning loading wheel)?")
  if conn_check=="1":
    print ("Got it. Let's fix that 'No Internet' error.")
    router_status= get_yes_no("Is your home wi-fi working on other devices, like your phone? (yes/no): ")
    if router_status=="no":
      print ("Step: Your router might be offline. Unplug your router for 30 seconds, then plug it back in.")
      resolved= get_yes_no("Did restarting the router fix the internet? (yes/no): ")
      if resolved=="no":
        print ("Escalation: We recommend contacting your Internet Service Provider.")
      else:
        print ("Awesome! We're back online")
    else:
      print ("Step: Go to Settings > Network on your streaming device and forget, then reconnect to your wi-fi.")
      resolved= get_yes_no("Did reconnecting your wi-fi work? (yes/no): ")
      if resolved=="no":
        print ("Escalation: Creating a ticket and routing you to a Network Specialist to check your device's MAC address...")
      else:
        print ("Perfect! Network settings refreshed successfully.")
  elif conn_check=="2":
    print ("Let's tackle that buffering issue.")
    placement_check= get_yes_no("Is your streaming device hidden inside a closed cabinet or behind a heavy TV? (yes/no): ")
    if placement_check=="yes":
      print ("Step: Move the device to an open area. Heavy objects can block the wi-fi signal and cause buffering.")
      resolved= get_yes_no("Did relocating the device stop the buffering? (yes/no): ")
      if resolved=="no":
        print ("Escalation: Creating a ticket and routing to Tier 2 Support to discuss a wi-fi range extender or ethernet adapter...")
      else:
        print ("Wonderful! Clearing the line of sight always helps.")
    else:
      print ("Step: Try lowering the video quality settings in the app, or restart your device to clear it's memory.")
      resolved= get_yes_no ("Did checking settings and restarting fix it? (yes/no): ")
      if resolved=="no":
        print ("Escalation: Creating ticket and connecting you to a Streaming Media Specialist now...")
      else:
        print ("Awesome! Back to smooth streaming.")
  else:
    print ("Invalid option. Redirecting you back to the main network settings menu.")
# This runs if they choose "3" (Audio/Video Issue) from the main menu
elif main_menu=="3":
  print ("Routing to Audio/Video Support")
  print ("Let's get your picture and sound working...")
  av_check= input ("Are you facing a 1. Black/Blank screen or 2. No sound at all?")
  if av_check=="1":
    print ("Let's troubleshoot that blank/black screen.")
    hdmi_check= get_yes_no("Is the HDMI cable plugged in all the way, or have you tried a different HDMI port? (yes/no): ")
    if hdmi_check=="no":
      print ("Step: Push the HDMI cable firmly into both the TV and the streaming device, or try switching to another HDMI port.")
      resolved= get_yes_no("Did securing or swapping the HDMI port fix the video? (yes/no): ")
      if resolved=="no":
        print ("Escalation: Creating a ticket and connecting you to our Hardware Team for deeper troubleshooting...")
      else:
        print ("Perfect! A loose HDMI connection is super common.")
    else:
      print ("Step: Try a different HDMI cable entirely and change your TV input source using your remote.")
      resolved= get_yes_no("Did changing the cables or input fix the screen? (yes/no): ")
      if resolved=="no":
        print ("Escalation: Creating a ticket and connecting you to our Hardware Team for video output failure...")
      else:
        print ("Awesome! Glad to see your screen is back!")
  elif av_check=="2":
    print ("Let's check your sound.")
    mute_check= get_yes_no("Is your TV or soundbar accidentally muted, or is the volume set below 5? (yes/no): ")
    if mute_check=="yes":
      print ("Step: Unmute your audio device and turn the volume up to 15 to test.")
      resolved= get_yes_no("Did unmuting and turning up the volume work? (yes/no): ")
      if resolved=="no":
        print ("Escalation: Creating a ticket and getting you connected to our Hardware Team...")
      else:
        print ("Great! Simple fixes are the best fixes.")
    else:
      print ("Step: Go to Settings > Display and Sound on your streaming device and change the audio output to 'Stereo'or 'Auto'. ")
      resolved= get_yes_no("Did changing the audio output format fix the sound? (yes/no): ")
      if resolved=="no":
        print ("Escalation: Creating ticket and routing you to Tier 2 Support for audio hardware configuration...")
      else:
        print ("Fantastic! Rock & Roll!")
  else:
    print ("Invalid option. Redirecting you back to the main audio/video menu.")
# This runs if they choose "4" (Remote Issue) from the main menu
elif main_menu=="4":
  print ("Routing to Remote Support...")
  print ("Let's troubleshoot your remote.")
  remote_check= input("Is the issue 1. Remote completely unresponsive, or 2. Voice control features not working?")
  if remote_check=="1":
    print ("Let's look into why the remote isn't responding.")
    battery_check= get_yes_no("Have you replaced the batteries with brand new ones within the last 5 minutes? (yes/no): ")
    if battery_check=="no":
      print ("Step: Remove the old batteries, press every button on the remote once to drain leftover power and insert fresh batteries.")
      resolved= get_yes_no("Did fresh batteries fix the remote responsiveness? (yes/no): ")
      if resolved=="no":
        print ("Escalation: Creating a ticket and connecting you to our Replacement Team to order a new remote...")
      else:
        print ("Awesome! Brand new batteries do the trick most of the time.")
    else:
      print ("Step: Press and hold the 'Home' button on your remote for 10-15 seconds right next to the streaming device to force a re-pair.")
      resolved= get_yes_no("Did forcing the remote to re-pair fix the issue? (yes/no): ")
      if resolved=="no":
        print ("Escalation: Creating a ticket and routing to Hardware Tech to check for a faulty IR/Bluetooth receiver on the device.")
      else:
        print ("Perfect! Your remote is re-paired and ready.")
  elif remote_check=="2":
    print ("Let's check your voice control features.")
    mic_permission= get_yes_no ("Is your streaming device logged into a valid account and connected to the internet? (yes/no): ")
    if mic_permission=="no":
      print ("Step: Voice commands require an active internet connection. Please go to the Network settings and sign in.")
      resolved= get_yes_no("Did connecting and signing in fix the voice control? (yes/no): ")
      if resolved=="no":
        print ("Escalation: Creating a ticket and connecting you to Account Services to verify your profile settings...")
      else:
        print ("Great! Voice features are officially online.")
    else:
      print ("Step: Go to Settings > Remotes & Accessories, unpair the remote, restart the streaming device, and pair the remote again.")
      resolved= get_yes_no ("Did unpairing and re-pairing clear the voice software glitch? (yes/no): ")
      if resolved=="no":
        print ("Escalating: Creating a ticket and transferring to Tier 2 Support to check for a corrupted update...")
      else:
        print ("Fantastic! Your microphone is working flawlessly now.")
  else:
    print ("Invalid option. Redirecting you back to the main remote menu.")

else:
  print ("Invalid choice. Please enter a number between 1 and 4.")
```
